## Objective 4: Real-World Data Science Application
In Data Science, we often receive raw data (like customer purchases) that needs to be aggregated and analyzed. This script simulates processing raw sales data using core Python fundamentals.




```python
# 1. Variables, Lists, and Dictionaries
print("--- Step 1: Loading Raw Data ---")

# A List containing Dictionaries (Simulating a dataset of customer purchases)
sales_data = [
    {"customer_id": 101, "item": "Laptop", "price": 1200.00, "quantity": 1},
    {"customer_id": 102, "item": "Mouse", "price": 25.50, "quantity": 2},
    {"customer_id": 103, "item": "Keyboard", "price": 45.00, "quantity": 1},
    {"customer_id": 104, "item": "Monitor", "price": 300.00, "quantity": 2}
]

# Variable for total number of records
total_records = len(sales_data)
print(f"Successfully loaded {total_records} sales records.\n")
```

```python
# 2. Functions and Loops
print("--- Step 2: Analyzing Data ---")

def calculate_total_revenue(data):
    """This function loops through the sales data to calculate total revenue."""
    total_revenue = 0.0

    # Loop through each record in our list
    for record in data:
        # Calculate revenue for this specific record (price * quantity)
        record_revenue = record["price"] * record["quantity"]
        total_revenue += record_revenue # Add to the grand total

    return total_revenue

# Calling the function and storing the result in a variable
total_sales_revenue = calculate_total_revenue(sales_data)
print(f"Total Revenue Generated: ${total_sales_revenue:,.2f}\n")
```

```python
# 3. File Handling
print("--- Step 3: Saving the Report ---")

# Writing our findings to a text file
file_name = "sales_summary_report.txt"

with open(file_name, "w") as file:
    file.write("--- CODOMAX SALES SUMMARY REPORT ---\n")
    file.write(f"Total Records Processed: {total_records}\n")
    file.write(f"Total Revenue Generated: ${total_sales_revenue:,.2f}\n")
    file.write("Status: Data processing complete.\n")

print(f"Report successfully saved as '{file_name}'.")

# Reading the file back to verify its contents
print("\n--- Verifying File Contents ---")
with open(file_name, "r") as file:
    print(file.read())
```
