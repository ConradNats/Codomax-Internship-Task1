# 📊 Exploratory Data Analysis & A/B Testing Workflow

This notebook demonstrates how to load, clean, and analyze a customer sales dataset containing missing values (`NaN`s) to evaluate the performance of an **A/B Test** across different customer segments.

---

## 🔍 Phase 1: Data Loading & Initial Inspection
We begin by importing our essential libraries (**Pandas** and **NumPy**) and loading the dataset. Initial inspection helps us understand the structure of the data and identify the gaps we need to clean.
```

````python
# Objective 1, 2 & 3: Pandas, NumPy, and Exploring Data
import pandas as pd
import numpy as np

# Load the dataset we just uploaded
df = pd.read_csv('Sales_with_NaNs_v1.3.csv')

print("--- 1. First 5 Rows of Raw Data ---")
display(df.head())

````

## 🛠️ Phase 2: Identifying Missing Data
Before performing any analysis, we need to locate any missing information (`NaN` values). This step counts how many blank entries exist in each column so we can plan our cleaning strategy.
```

````python
#FINDING MISSING VALUES
print("\n--- 2. Finding Missing Data ---")
# .isnull().sum() is a great way to see exactly how many NaNs are in each column
missing_data = df.isnull().sum()
display(missing_data)
````

## 🧼 Phase 3: Data Cleaning & Imputation

To ensure our analysis is accurate, we must handle missing values rather than simply discarding them:

1. **Numerical Data**: For columns containing continuous values (like Sales and Satisfaction levels), we fill missing fields with the column's **Average (Mean)**. This keeps our overall statistics stable without skewing the results.
```

````python
# Objective 4: Handle missing values

# CLEANING NUMBERS: Fill missing sales & satisfaction data with the column's Average (Mean)
num_columns = ['Sales_Before', 'Sales_After', 'Customer_Satisfaction_Before', 'Customer_Satisfaction_After']
for col in num_columns:
    df[col] = df[col].fillna(df[col].mean())
print(".....***Successfully cleaned***.....")
````

2. **Categorical Data**: For descriptive columns (like Group, Segment, or Purchase status), we fill missing fields using the **Most Common Value (Mode)**.

Let's verify that all missing values have been successfully handled!
```

````python
# CLEANING TEXT: Fill missing categorical data with the most common value (Mode)
text_columns = ['Group', 'Customer_Segment', 'Purchase_Made']
for col in text_columns:
    df[col] = df[col].fillna(df[col].mode()[0])

print("✅ Data successfully cleaned! Let's check for missing values again:")
display(df.isnull().sum()) # Should all be 0 now!
````

## 📈 Phase 4: Feature Engineering

To measure customer improvement, we create a new metric: **Sales Growth**.
This is calculated by subtracting `Sales_Before` from `Sales_After` for each individual customer, giving us a direct look at the change in behavior.
```

````python
# Let's create a new column to see how much Sales grew for each customer
df['Sales_Growth'] = df['Sales_After'] - df['Sales_Before']
print("--- 3. New Column Created ---")
df[['Sales_Before', 'Sales_After', 'Sales_Growth']]
````

## 📊 Phase 5: A/B Test & Segment Analysis

Now that our data is clean and structured, we can answer key business questions:
1. **A/B Test Evaluation**: Does the **Treatment** group show higher average sales and growth compared to the **Control** group?
2. **Customer Segmentation**: Which customer segments generated the highest overall sales after the experiment?
```

````python
# Objective 4: Apply basic data analysis techniques
print("--- 4. Analyzing A/B Test Results ---")

# Group by 'Control' vs 'Treatment' to see which group performed better!
group_analysis = df.groupby('Group')[['Sales_Before', 'Sales_After', 'Sales_Growth']].mean().reset_index()

print("\n📊 Average Sales Performance by Group:")
display(group_analysis)

# Group by Customer Segment
segment_analysis = df.groupby('Customer_Segment')['Sales_After'].sum().reset_index()
print("\n📊 Total Sales After by Customer Segment:")
display(segment_analysis)
````
